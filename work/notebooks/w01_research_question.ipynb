## The Question

### Research Question

Which content pages should be reviewed first because they show signs of declining performance while still receiving meaningful search visibility?

### Decision

The goal is to decide which content pages should be placed at the top of the review queue.

### Unit of Analysis

Each row in the dataset represents one content page.

### Output

The final output will be a ranked list of content pages with opportunity scores and reason codes explaining why each page should be reviewed.

### Action

A content manager or SEO specialist can review the highest-ranked pages and decide whether to refresh, expand, rewrite, monitor, or leave the content unchanged.

### Cost of a Wrong Recommendation

If low-priority pages are recommended, valuable review time may be wasted. If important declining pages are missed, they may continue losing traffic and visibility, reducing the effectiveness of the review process.

### Why Machine Learning Helps

Machine learning can combine several search and engagement signals at the same time instead of relying on a single rule. This helps rank pages more accurately and supports better decision-making.

## Quick Look at the Data

The starter dataset contains **30,000** content pages.

The average number of impressions during the last 90 days is **5,200.37**, while the average number of sessions is **37.07**.

Approximately **54.2%** of the pages have a **down** trend direction. This suggests that a large portion of the content may require review, making Refresh / Content Opportunity Scoring a useful research direction for the remaining internship weeks.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the uploaded dataset
df = pd.read_csv("content_refresh_anonymized.csv")

# Calculate statistics
total_pages = len(df)
avg_impressions = df["impressions_90d"].mean()
avg_sessions = df["sessions_90d"].mean()

down_pages = (df["trend_direction"] == "down").sum()
down_percentage = (down_pages / total_pages) * 100

# Display results
print(f"Total pages: {total_pages}")
print(f"Average impressions (90d): {avg_impressions:.2f}")
print(f"Average sessions (90d): {avg_sessions:.2f}")
print(f"Pages with 'down' trend: {down_pages}")
print(f"Percentage of 'down' pages: {down_percentage:.2f}%")
```
