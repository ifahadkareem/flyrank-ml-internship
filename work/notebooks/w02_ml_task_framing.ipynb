## 3. Success metric

*One metric you can defend. What number means 'good'?*

My primary success metric is Precision@20.

Precision@20 measures how many of the first 20 recommended pages are judged by a human reviewer to be genuine content-refresh opportunities.

This metric matches the real action because a content manager will usually review only the highest-ranked pages rather than all 30,000 pages.

I will provisionally consider Precision@20 of at least 0.75 to be useful. This means that at least 15 of the first 20 recommendations should be relevant.

A false positive wastes review time. A false negative may allow an important declining page to continue losing visibility.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

##5. Why ML beats a fixed rule here
A simple fixed-rule baseline could flag every page with a downward trend and more than a chosen number of impressions.

This rule would be useful as a baseline, but it would not be enough by itself. In the starter dataset, 16,262 of 30,000 pages, or approximately 54.21%, have a downward trend. Flagging every declining page would create an excessively large review queue.

Refresh priority may depend on several interacting signals, including impressions, clicks, sessions, CTR, average position, engagement, content age, time since the last update, and recent performance changes.

A single threshold cannot easily represent all combinations of these signals. An ML scoring method may learn more complex patterns and produce a more useful ordering of pages.

The output will remain decision support. A human reviewer will make the final decision about whether each page should be refreshed, rewritten, expanded, monitored, or left unchanged.
